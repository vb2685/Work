## Question 1: Markdown + HTML Example

Below is an example of mixing Markdown and HTML in Jupyter Notebook:

<div style="background-color:black; color:white; padding:10px; border-radius:5px;">
<h3>Euler's Formula</h3>
<p><i>e<sup>iθ</sup> = cos(θ) + i·sin(θ)</i></p>
</div>


```python
import numpy as np
import matplotlib.pyplot as plt

def eulerPlot():
  #Define the domain x (e.g., sweeping across 3 full rotations)
  x = np.linspace(0, 10 * np.pi, 1000)

  #Compute Real and Imaginary components
  y_real = np.cos(x)
  z_imag = np.sin(x)

  #Create the 3D plot
  fig = plt.figure(figsize=(10, 7), facecolor='black')
  ax = fig.add_subplot(111, projection='3d', facecolor= 'black')

  #Plot the 3D helix
  ax.plot(x, y_real, z_imag, label=r'$e^{ix}$ Helix', color='yellow', linewidth=2.5)

  #Draw reference planes or axis lines for visual grounding
  ax.plot(x, np.zeros_like(x), np.zeros_like(x), color='blue', linestyle='--', alpha=0.5)

  #Label the axes clearly
  ax.set_xlabel('Angle/Time ($x$)', fontsize=11, labelpad=10, color='white')
  ax.set_ylabel('Real Axis: $\\cos(x)$', fontsize=11, labelpad=10, color='white')
  ax.set_zlabel('Imaginary Axis: $\\sin(x)$', fontsize=11, labelpad=10, color='white')
  ax.set_title("Euler's Formula Unrolled as a 3D Helix", fontsize=14, pad=20, color ='white')

  ax.legend(loc='upper right')
  plt.show()

#Execute Euler plot and perform time analysis
%time eulerPlot()
```

## Question 2: Latex Example
Plot 3phase voltages with following equation:
\begin{align} V_a(t) &= V_m \sin(\omega t) \\
V_b(t) &= V_m \sin\left(\omega t - \frac{2\pi}{3}\right) \\
V_c(t) &= V_m \sin\left(\omega t + \frac{2\pi}{3}\right)
\end{align}


```python
import numpy as np
import matplotlib.pyplot as plt

def plot_3phase_Voltages() :
   # Parameters
    f = 50
    Vm = 325
    t = np.linspace(0, 0.04, 1000)

    omega = 2 * np.pi * f

    # Three-phase voltages
    Va = Vm * np.sin(omega * t)
    Vb = Vm * np.sin(omega * t - 2*np.pi/3)
    Vc = Vm * np.sin(omega * t + 2*np.pi/3)

    # Plot
    fig, ax = plt.subplots(figsize=(12, 5))

    ax.plot(t * 1000, Va, 'r', lw=2, label='Phase A')
    ax.plot(t * 1000, Vb, 'g', lw=2, label='Phase B')
    ax.plot(t * 1000, Vc, 'b', lw=2, label='Phase C')

    ax.set_title("Three-Phase Voltages")
    ax.set_xlabel("Time (ms)")
    ax.set_ylabel("Voltage (V)")
    ax.grid(True)
    ax.legend()
    plt.show()
%time plot_3phase_Voltages()
```
